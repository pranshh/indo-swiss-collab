```python
from tqdm import tqdm

# Set tqdm to work with groupby
tqdm.pandas()

# Group manually with tqdm
print("👥 Grouping authors with progress...")
grouped = []
for work_id, group in tqdm(author.groupby('work_id'), total=author['work_id'].nunique()):
    names = '; '.join(group['display_name'].dropna().unique())
    ids = '; '.join(group['author_id'].dropna().unique())
    grouped.append({
        'work_id': work_id,
        'author_names': names,
        'author_ids': ids
    })

# Convert to DataFrame
grouped_authors = pd.DataFrame(grouped)

# Merge
print("🔗 Merging grouped authors into publications...")
df_final = publication.merge(grouped_authors, on='work_id', how='left')

print("✅ All done, boss.")
```

```python
import json

def extract_topic_names(topic_str):
    try:
        topic_list = json.loads(topic_str)
        return '; '.join(set([entry['display_name'] for entry in topic_list if 'display_name' in entry]))
    except:
        return None

print("🧠 Extracting display_name from topics...")
df_final['topic_names'] = df_final['topics'].progress_apply(extract_topic_names)
```
